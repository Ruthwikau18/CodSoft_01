**CODSOFT TASK 1 - DATA CLEANING & PREPROCESSING**

```python
# 1. Import libraries
import pandas as pd
import numpy as np
```

```python
# 2. Load dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print("Dataset loaded successfully!")
```

```python
# 3. Inspect dataset
print("\nFirst 10 rows:")
display(df.head(10))

print("\nFirst 10 columns:")
display(df.tail(10))

print("\nDataset shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python
# 4. Check missing values

# Missing value percentage
missing_data = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': df.isnull().mean() * 100
})

print("\nMissing value analysis:")
display(missing_data)
```

```python
# 5. Check duplicates
print("\nDuplicate records:")
print(df.duplicated().sum())
```

```python
# 6. Check inconsistent/categorical data
print("\nUnique Sex values:")
print(df['Sex'].unique())

print("\nSex value counts:")
print(df['Sex'].value_counts())

print("\nUnique Embarked values:")
print(df['Embarked'].unique())

print("\nEmbarked value counts:")
print(df['Embarked'].value_counts())
```

```python
# 7. Handle missing Age values
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# 8. Handle missing Embarked values
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
# 9. Handle missing Cabin values
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
# 10. Remove duplicate records
df = df.drop_duplicates()
```

```python
# 11. Correct data types
df['PassengerId'] = df['PassengerId'].astype(int)
df['Survived'] = df['Survived'].astype(int)
df['Pclass'] = df['Pclass'].astype(int)
df['Age'] = df['Age'].astype(float)
df['Fare'] = df['Fare'].astype(float)
```

```python
# 12. Standardize text data
df['Sex'] = df['Sex'].str.strip().str.lower()
df['Embarked'] = df['Embarked'].str.strip().str.upper()
# 13. Final validation
print("\n========== FINAL DATASET ==========")

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate records after cleaning:")
print(df.duplicated().sum())

print("\nFinal data types:")
print(df.dtypes)

print("\nFinal dataset shape:")
print(df.shape)

print("\nCleaned dataset:")
display(df.head(10))
```

```python
# 14. Save cleaned dataset
output_file = 'Cleaned_titanic_dataset.csv'
df.to_csv(output_file, index=False)

print(f"\nCleaned dataset saved as: {output_file}")
```
